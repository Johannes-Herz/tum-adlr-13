## **Generation**
---

```python
circle_config: config.CircleConfig = config.CircleConfig(num_samples=500)
rectangle_config: config.RectangleConfig = config.RectangleConfig(num_samples=500)
triangle_config: config.TriangleConfig = config.TriangleConfig(num_samples=500, min_height=0.5)
grid_config: config.GridConfig = config.GridConfig(resolution=64)
rng: np.random.Generator = np.random.default_rng(seed=42)
```

```python
circle_sample_fn: sampling.SampleFunction = sampling.REGISTRY[circle_config.primitive]
circle_sdf_builder: sdfs.SignedDistanceFunctionBuilder = sdfs.REGISTRY[circle_config.primitive]
circle_occupancy_grids, circle_sdf_grids, circle_sample_params = generation.generate(primitive_config=circle_config, grid_config=grid_config, sdf_builder=circle_sdf_builder, sample_fn=circle_sample_fn, rng=rng)
```

```python
rectangle_sample_fn: sampling.SampleFunction = sampling.REGISTRY[rectangle_config.primitive]
rectangle_sdf_builder: sdfs.SignedDistanceFunctionBuilder = sdfs.REGISTRY[rectangle_config.primitive]
rectangle_occupancy_grids, rectangle_sdf_grids, rectangle_sample_params = generation.generate(primitive_config=rectangle_config, grid_config=grid_config, sdf_builder=rectangle_sdf_builder, sample_fn=rectangle_sample_fn, rng=rng)
```

```python
triangle_sample_fn: sampling.SampleFunction = sampling.REGISTRY[triangle_config.primitive]
triangle_sdf_builder: sdfs.SignedDistanceFunctionBuilder = sdfs.REGISTRY[triangle_config.primitive]
triangle_occupancy_grids, triangle_sdf_grids, triangle_sample_params = generation.generate(primitive_config=triangle_config, grid_config=grid_config, sdf_builder=triangle_sdf_builder, sample_fn=triangle_sample_fn, rng=rng)
```

## **Visualization**
---

```python
K: int = 9
show(circle_occupancy_grids, k=K, rng=rng)
show(rectangle_occupancy_grids, k=K, rng=rng)
show(triangle_occupancy_grids, k=K, rng=rng)
```
