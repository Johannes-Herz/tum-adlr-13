## **Generation**
---

```python
RESOLUTION: int = 128
N: int = 100
SEED: int = 100
RNG: np.random.Generator = np.random.default_rng(seed=42)
```

```python
# occupancy_grids = np.load(file=rf"C:\dev\tum-adlr-13\data\generated\rectangle\res{RESOLUTION}\v001\occupancy.npy")
occupancy_grids = np.load(r"C:\dev\tum-adlr-13\data\full_grid\shapenet\02773838_occupancy_grid.npz")["masks"]

occupancy_grids = occupancy_grids[:64, ...]
B, H, W = occupancy_grids.shape
print("Occupancy Grids Shape", occupancy_grids.shape)
```

```python
rays = raycasting.sample(resolution=RESOLUTION, b=B, n=N, rng=RNG)
print("Rays Shape", rays.shape)
```

```python
traversals = raycasting.traverse(rays=rays, resolution=RESOLUTION)
_, N, M, _ = traversals.shape
print("Traversals Shape", traversals.shape)
```

```python
num_rays: np.ndarray = RNG.integers(low=N - 1, high=N, size=(B, 1))
active_rays: np.ndarray = np.arange(N) < num_rays  # (B, N)
traversal_grids = raycasting.cast(occupancy_grids=occupancy_grids, traversals=traversals, num_rays=num_rays)
print("Traversal Grids Shape", traversal_grids.shape)
```

## **Visualization**

```python
K: int = 9
ray_vis.show(rays=rays, resolution=RESOLUTION, k=K)
traversal_vis.show(traversals=traversals, resolution=RESOLUTION, rays=rays, k=K)
traversal_grid_vis.show(traversal_grids=traversal_grids, occupancy_grids=occupancy_grids, k=K)
```
